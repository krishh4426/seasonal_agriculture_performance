### Box Plots for Key Numerical Variables

```python
key_numeric_variables = [
    'Rainfall_mm',
    'Yield_Tonnes_Ha',
    'Fertilizer_kg_ha',
    'Water_Used_m3',
    'Total_Cost_INR',
    'Revenue_INR',
    'Profit_INR'
]

for col in key_numeric_variables:
    plt.figure(figsize=(10, 6))
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

```python
outlier_summary = []

for col in key_numeric_variables:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        'Variable': col,
        'Lower_Bound': lower,
        'Upper_Bound': upper,
        'Outlier_Count': count,
        'Outlier_Percentage': count / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Correlation Analysis

First, let's look at the correlation matrix for all numerical columns to understand the linear relationships between different variables. A heatmap is a great way to visualize this matrix.

```python
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(20, 15))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```

From the correlation heatmap, we can observe various relationships:

*   **Positive correlations** (closer to 1, warm colors) indicate that as one variable increases, the other tends to increase as well.
*   **Negative correlations** (closer to -1, cool colors) indicate that as one variable increases, the other tends to decrease.
*   **Near-zero correlations** (closer to 0, neutral colors) suggest little to no linear relationship between the variables.

Key areas to focus on would be correlations involving `Yield_Tonnes_Ha`, `Production_Tonnes`, and `Profit_INR` with environmental factors (`Rainfall_mm`, `Avg_Temperature_C`, `Humidity_pct`, `Sunlight_Hours_Day`, `Soil_pH`, `Soil_Moisture_pct`, `Disease_Pest_Risk_pct`) and farming practices (`Farm_Area_Hectares`, `Nitrogen_kg_ha`, `Phosphorus_kg_ha`, `Potassium_kg_ha`, `Fertilizer_kg_ha`, `Pesticide_Litre_ha`, `Seed_Quality_Score`, `Water_Used_m3`, `Water_Efficiency_t_per_1000m3`).

This heatmap provides a comprehensive overview of how different aspects of agricultural activities and environmental conditions are related, which is crucial for identifying meaningful patterns, trends, relationships, and variations as stated in your problem statement.

### Seasonal Comparison of Agricultural Performance

To get a clearer picture of how agricultural performance varies across seasons, let's calculate the mean of key metrics for each season.

```python
import matplotlib.pyplot as plt
import seaborn as sns

seasonal_performance = df.groupby('Season')[['Yield_Tonnes_Ha', 'Production_Tonnes', 'Profit_INR']].mean().reset_index()
display(seasonal_performance)
```

Now, let's visualize the average `Yield_Tonnes_Ha` by season.

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Season', y='Yield_Tonnes_Ha', data=seasonal_performance)
plt.title('Average Yield (Tonnes/Ha) by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')

plt.show()
```

Next, we visualize the average `Production_Tonnes` by season.

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Season', y='Production_Tonnes', data=seasonal_performance)
plt.title('Average Production (Tonnes) by Season')
plt.xlabel('Season')
plt.ylabel('Average Production (Tonnes)')

plt.show()
```

Finally, let's look at the average `Profit_INR` by season.

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Season', y='Profit_INR', data=seasonal_performance)
plt.title('Average Profit (INR) by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')

plt.show()
```
